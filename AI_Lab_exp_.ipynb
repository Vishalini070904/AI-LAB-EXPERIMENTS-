```python
#EXP 1
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'E'],
    'D': ['B', 'G'],
    'E': ['B', 'C', 'F'],
    'F': ['E', 'H'],
    'G': ['D', 'H'],
    'H': ['F', 'G']
}


def bfs(graph, start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}
    distance = {start: 0}
    order = []

    print("BFS: Hospital Navigation")
    print(f"Start: {start} | Goal: {goal}\n")

    while queue:
        node = queue.popleft()
        order.append(node)

        print(f"Step {len(order)}: Node = {node}")
        print(f"  Queue: {list(queue)}")

        if node == goal:
            print("  Goal reached!")
            break

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = node
                distance[neighbor] = distance[node] + 1
                queue.append(neighbor)
                print(f"  {node} -> {neighbor}: Added to queue")
            else:
                print(f"  {node} -> {neighbor}: Already visited")

        print(f"  Visited: {sorted(visited)}")

    if goal not in parent:
        print("\nNo route found.")
        return

    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("\n--- Final Result ---")
    print("Traversal order:", " -> ".join(order))
    print("Shortest path  :", " -> ".join(path))
    print("Path cost      :", distance[goal], "corridors")
    print("Nodes explored :", len(order))


bfs(graph, 'A', 'H')
```

```python
#EXP 2
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['H'],
    'F': ['H'],
    'G': ['H'],
    'H': ['L'],
    'L': []
}


def dfs(graph, start, goal):
    visited = set()
    path = []
    order = []

    def search(node):
        visited.add(node)
        order.append(node)
        path.append(node)

        print(f"Step {len(order)}: Node = {node}")
        print(f"  Current path: {' -> '.join(path)}")

        if node == goal:
            print("  Goal reached!")
            return True

        for neighbor in graph[node]:
            if neighbor not in visited:
                print(f"  Exploring edge: {node} -> {neighbor}")

                if search(neighbor):
                    return True
            else:
                print(f"  Skipping {neighbor}: Already visited")

        path.pop()
        print(f"  Backtracking from {node}")
        return False

    print("DFS: AI Pipeline Reachability")
    print(f"Start: {start} | Goal: {goal}\n")

    found = search(start)

    print("\n--- Final Result ---")
    print("Traversal order:", " -> ".join(order))

    if found:
        print("Path found     :", " -> ".join(path))
        print("Path length    :", len(path) - 1, "edges")
    else:
        print("No path exists.")
        print("Path found     : None")

    print("Nodes explored :", len(order))


dfs(graph, 'A', 'H')
```

```python
#EXP 3
from collections import deque

def water_jug_bfs():
    start = (0, 0)
    goal = 2
    queue = deque([start])
    visited = {start}
    parent = {start: None}
    action = {start: "Initial state"}
    capacities = (4, 3)

    def get_next_states(a, b):
        states = []

        states.append(((4, b), "Fill 4-litre jug"))
        states.append(((a, 3), "Fill 3-litre jug"))
        states.append(((0, b), "Empty 4-litre jug"))
        states.append(((a, 0), "Empty 3-litre jug"))

        transfer = min(a, capacities[1] - b)
        states.append(((a - transfer, b + transfer),
                       "Pour 4-litre jug into 3-litre jug"))

        transfer = min(b, capacities[0] - a)
        states.append(((a + transfer, b - transfer),
                       "Pour 3-litre jug into 4-litre jug"))

        return states

    step = 0
    goal_state = None

    print("Water Jug Problem: State-Space Search")
    print("Start state:", start)
    print("Goal: Exactly 2 litres in the 4-litre jug\n")

    while queue:
        current = queue.popleft()
        step += 1
        a, b = current

        print(f"Step {step}: State = {current}")
        print("  Queue:", list(queue))

        if a == goal:
            goal_state = current
            print("  Goal reached!")
            break

        for next_state, operation in get_next_states(a, b):
            if next_state not in visited:
                visited.add(next_state)
                parent[next_state] = current
                action[next_state] = operation
                queue.append(next_state)

                print(f"  Action: {operation}")
                print(f"  New state: {next_state}")
            else:
                print(f"  Skipped {next_state}: Already visited")

    print("\n--- Final Result ---")

    if goal_state is None:
        print("No solution found.")
        return

    path = []
    current = goal_state

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    for i, state in enumerate(path):
        print(f"Step {i}: {state} | {action[state]}")

    print("\nSolution path:", " -> ".join(map(str, path)))
    print("Operations required:", len(path) - 1)
    print("States discovered:", len(visited))


water_jug_bfs()
```

```python
#EXP 4
import heapq

# Weighted graph from the experiment image

graph = {
    'A': [('B', 4), ('C', 3)],
    'B': [('A', 4), ('C', 2), ('D', 5)],
    'C': [('A', 3), ('B', 2), ('D', 6), ('E', 4)],
    'D': [('B', 5), ('C', 6), ('E', 1), ('F', 2)],
    'E': [('C', 4), ('D', 1), ('F', 3)],
    'F': [('D', 2), ('E', 3)]
}

# Estimated cost from each node to goal F

h = {
    'A': 6,
    'B': 4,
    'C': 5,
    'D': 2,
    'E': 3,
    'F': 0
}

def astar(start, goal):

    # Priority queue entries: (f, node, g, path)
    open_list = [(h[start], start, 0, [start])]
    best_g = {start: 0}
    expanded = []

    print("A* Search: Intelligent Route Planning")
    print(f"Start: {start} | Goal: {goal}\n")

    while open_list:
        f, node, g, path = heapq.heappop(open_list)

        # Ignore outdated entries
        if g != best_g.get(node):
            continue

        expanded.append(node)
        print(f"Step {len(expanded)}: Expand {node}")
        print(f"  g(n)={g}, h(n)={h[node]}, f(n)={f}")

        if node == goal:
            print("  Goal reached!")
            break

        for neighbor, cost in graph[node]:
            new_g = g + cost
            new_f = new_g + h[neighbor]

            if new_g < best_g.get(neighbor, float('inf')):
                best_g[neighbor] = new_g
                new_path = path + [neighbor]

                heapq.heappush(
                    open_list,
                    (new_f, neighbor, new_g, new_path)
                )

                print(
                    f"  {node}->{neighbor}: "
                    f"g={new_g}, h={h[neighbor]}, f={new_f} "
                    f"| Updated"
                )
            else:
                print(f"  {node}->{neighbor}: No better cost")

        pending = sorted(
            (item[1], item[2], item[0])
            for item in open_list
            if item[2] == best_g.get(item[1])
        )

        print("  OPEN (node, g, f):", pending, "\n")

    if node == goal:
        print("--- Final Result ---")
        print("Optimal path    :", " -> ".join(path))
        print("Total path cost :", g)
        print("Nodes expanded  :", len(expanded))
        print("Expansion order :", " -> ".join(expanded))
    else:
        print("No route found.")

astar('A', 'F')
```

```python
#EXP 5
N = 8
board = [-1] * N
attempts = 0


def is_safe(row, col):
    for r in range(row):
        if board[r] == col:
            return False

        if abs(board[r] - col) == abs(r - row):
            return False

    return True


def solve(row):
    global attempts

    if row == N:
        return True

    for col in range(N):
        attempts += 1

        if is_safe(row, col):
            board[row] = col

            if solve(row + 1):
                return True

            # Backtrack: remove the queen
            board[row] = -1

    return False


if solve(0):
    print("Solution found!")
    print("\nChessboard:")

    for row in range(N):
        for col in range(N):
            if board[row] == col:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

    print("\nQueen positions (Row, Column):")
    for row in range(N):
        print(f"Queen {row + 1}: ({row + 1}, {board[row] + 1})")

    print("\nTotal placement attempts:", attempts)
else:
    print("No solution exists.")
```

```python
#EXP 6
# Map Colouring using CSP and Backtracking

regions = ['A', 'B', 'C', 'D', 'E', 'F']

colours = ['Red', 'Green', 'Blue']

adjacency = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['C', 'D', 'F'],
    'F': ['D', 'E']
}

assignment = {}
attempts = 0


def is_safe(region, colour):
    for neighbour in adjacency[region]:
        if assignment.get(neighbour) == colour:
            return False
    return True


def solve(index):
    global attempts

    if index == len(regions):
        return True

    region = regions[index]

    for colour in colours:
        attempts += 1

        if is_safe(region, colour):
            assignment[region] = colour

            if solve(index + 1):
                return True

            # Backtrack
            del assignment[region]

    return False


if solve(0):
    print("Valid Colour Assignment:\n")

    for region in regions:
        print(region, ":", assignment[region])

    print("\nConstraint Verification:")

    valid = True
    for region in regions:
        for neighbour in adjacency[region]:
            if assignment[region] == assignment[neighbour]:
                print("Conflict:", region, neighbour)
                valid = False

    if valid:
        print("All adjacency constraints satisfied.")

    print("Total assignment attempts:", attempts)

else:
    print("No valid colouring exists.")
```

```python
#EXP 7
# Minimax Algorithm

tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [3, 5],
    'E': [2, 9],
    'F': [12, 5],
    'G': [23, 23]
}

values = {}


def minimax(node, maximizing):
    if isinstance(tree[node][0], int):
        if maximizing:
            values[node] = max(tree[node])
        else:
            values[node] = min(tree[node])
        return values[node]

    child_values = [
        minimax(child, not maximizing)
        for child in tree[node]
    ]

    if maximizing:
        values[node] = max(child_values)
    else:
        values[node] = min(child_values)

    return values[node]


result = minimax('A', True)

print("Evaluated Node Values:")
for node in sorted(values):
    print(node, ":", values[node])

best_move = max(tree['A'], key=lambda child: values[child])

print("\nBest Move for AI:", best_move)
print("Maximum Utility Value:", result)
```

```python
#EXP 8
# Alpha-Beta Pruning

tree = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F', 'G'],
    'C': ['H', 'I', 'J'],
    'D': ['K', 'L', 'M'],
    'E': [3, 5],
    'F': [6, 9],
    'G': [1, 2],
    'H': [4, 6],
    'I': [7, 8],
    'J': [9, 10],
    'K': [1, 3],
    'L': [12, 14],
    'M': [15, 18]
}

values = {}
nodes_evaluated = 0
branches_pruned = 0


def minimax(node, alpha, beta, maximizing):
    global nodes_evaluated, branches_pruned

    children = tree[node]

    # Terminal node: evaluate utility values
    if isinstance(children[0], int):
        return max(children) if maximizing else min(children)

    if maximizing:
        best = float('-inf')

        for i, child in enumerate(children):
            value = minimax(child, alpha, beta, False)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                branches_pruned += len(children) - i - 1
                break

    else:
        best = float('inf')

        for i, child in enumerate(children):
            value = minimax(child, alpha, beta, True)
            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                branches_pruned += len(children) - i - 1
                break

    values[node] = best
    return best


def count_terminal_nodes(node):
    global nodes_evaluated

    children = tree[node]

    if isinstance(children[0], int):
        for value in children:
            nodes_evaluated += 1
        return

    for child in children:
        count_terminal_nodes(child)


# Run Alpha-Beta Pruning
result = minimax('A', float('-inf'), float('inf'), True)

# Find values of the root's children for the selected move
print("Best Move Values:")

for child in tree['A']:
    values[child] = minimax(
        child, float('-inf'), float('inf'), False
    )
    print(child, ":", values[child])

# Count actual terminal evaluations during a single search
nodes_evaluated = 0
branches_pruned = 0

def alpha_beta(node, alpha, beta, maximizing):
    global nodes_evaluated, branches_pruned

    children = tree[node]

    if isinstance(children[0], int):
        best = float('-inf') if maximizing else float('inf')

        for i, value in enumerate(children):
            nodes_evaluated += 1

            if maximizing:
                best = max(best, value)
                alpha = max(alpha, best)
            else:
                best = min(best, value)
                beta = min(beta, best)

            if alpha >= beta:
                branches_pruned += len(children) - i - 1
                break

        return best

    best = float('-inf') if maximizing else float('inf')

    for i, child in enumerate(children):
        value = alpha_beta(child, alpha, beta, not maximizing)

        if maximizing:
            best = max(best, value)
            alpha = max(alpha, best)
        else:
            best = min(best, value)
            beta = min(beta, best)

        if alpha >= beta:
            branches_pruned += len(children) - i - 1
            break

    return best


values.clear()
best_value = alpha_beta('A', float('-inf'), float('inf'), True)

# Determine the best move from the minimax values
for child in tree['A']:
    values[child] = minimax(
        child, float('-inf'), float('inf'), False
    )

best_move = max(tree['A'], key=lambda x: values[x])

print("\nBest Move for AI:", best_move)
print("Maximum Utility Value:", best_value)
print("Terminal Nodes Evaluated:", nodes_evaluated)
print("Branches Pruned:", branches_pruned)
```

```python
#EXP 9
facts = {
    "fever",
    "cough",
    "high_temperature",
    "age_over_65"
}

rules = [
    (["fever", "cough"], "respiratory_symptoms"),
    (["respiratory_symptoms", "high_temperature"],
     "suspected_infection"),
    (["suspected_infection", "age_over_65"],
     "priority_consultation")
]

trace = []

print("Forward Chaining Using Facts and Rules\n")

print("Initial Facts:")
for fact in sorted(facts):
    print("-", fact)

changed = True

while changed:
    changed = False

    for conditions, conclusion in rules:
        if all(condition in facts for condition in conditions):
            if conclusion not in facts:
                facts.add(conclusion)
                changed = True
                trace.append(
                    f"IF {conditions} THEN {conclusion}"
                )
                print(f"\nRule applied: {trace[-1]}")
                print(f"New fact inferred: {conclusion}")

print("\n--- Final Facts ---")
for fact in sorted(facts):
    print("-", fact)

print("\n--- Final Conclusion ---")
if "priority_consultation" in facts:
    print("Patient qualifies for a priority consultation.")
else:
    print("Priority consultation criteria not established.")

print("\nTotal new facts inferred:", len(trace))
```

```python
#EXP 10
clauses = [
    {"EnjoysLearning"},                  # C1
    {"HasFreeTime"},                     # C2
    {"NotEnjoysLearning", "Happy"},      # C3
    {"NotHasFreeTime", "Happy"}          # C4
]

goal = "Happy"

print("Resolution Refutation\n")

print("--- Knowledge Base (CNF) ---")

for i, clause in enumerate(clauses, start=1):
    print(f"C{i}: {' OR '.join(sorted(clause))}")

# Negate the goal and add it to the knowledge base

negated_goal = "Not" + goal

clauses.append({negated_goal})

print(f"\nNegated goal added: {negated_goal}")

# Resolve complementary literals

derived = set()

empty_clause_found = False

print("\n--- Resolution Steps ---")

changed = True

while changed and not empty_clause_found:

    changed = False

    current_clauses = list(clauses)

    for i in range(len(current_clauses)):

        for j in range(i + 1, len(current_clauses)):

            c1 = current_clauses[i]

            c2 = current_clauses[j]

            for literal in c1:

                if literal.startswith("Not"):
                    complement = literal[3:]
                else:
                    complement = "Not" + literal

                if complement in c2:

                    resolvent = (c1 - {literal}) | (c2 - {complement})

                    print(
                        f"Resolve {sorted(c1)} and {sorted(c2)} "
                        f"on {literal} -> {sorted(resolvent)}"
                    )

                    if not resolvent:
                        empty_clause_found = True
                        break

                    frozen_resolvent = frozenset(resolvent)

                    if (frozen_resolvent not in
                            {frozenset(c) for c in clauses} and
                        frozen_resolvent not in derived):

                        derived.add(frozen_resolvent)

                        changed = True

            if empty_clause_found:
                break

        if empty_clause_found:
            break

    for clause in derived:
        new_clause = set(clause)

        if new_clause not in clauses:
            clauses.append(new_clause)

    derived.clear()

print("\n--- Final Result ---")

if empty_clause_found:
    print("Empty clause derived: Yes")
    print("Contradiction found.")
    print("Therefore, the goal 'Happy' is proved.")

else:
    print("Empty clause derived: No")
    print("The goal could not be proved by this resolution run.")
```

```python
#EXP 11
parents = [
    ("Ravi", "Arun"),
    ("Ravi", "Meena"),
    ("Arun", "Kiran"),
    ("Meena", "Priya"),
    ("Suresh", "Ravi")
]

def is_parent(parent, child):
    return (parent, child) in parents

def get_children(parent):
    return [child for p, child in parents if p == parent]

def get_parents(child):
    return [parent for parent, c in parents if c == child]

def get_grandparents(person):
    grandparents = []

    for parent in get_parents(person):
        grandparents.extend(get_parents(parent))

    return grandparents

# Display family relationships

person = "Kiran"

print("Family Tree Relationships")
print("-------------------------")

print("Parents of", person, ":", get_parents(person))

print("Children of Ravi:", get_children("Ravi"))

print("Grandparents of", person, ":", get_grandparents(person))

print("\nIs Ravi a parent of Arun?", is_parent("Ravi", "Arun"))

print("Is Suresh a grandparent of Kiran?",
      "Suresh" in get_grandparents("Kiran"))
```

```python
#EXP 12
import re

# Healthcare symptom patterns

symptom_patterns = {

    "Emergency Warning Signs": [
        r"\bchest pain\b",
        r"\bdifficulty breathing\b",
        r"\bshortness of breath\b",
        r"\bface drooping\b",
        r"\bsudden weakness\b",
        r"\bsevere bleeding\b"
    ],

    "Respiratory Symptoms": [
        r"\bcough\b",
        r"\bsore throat\b",
        r"\brunny nose\b",
        r"\bfever\b"
    ],

    "Digestive Symptoms": [
        r"\bnausea\b",
        r"\bvomiting\b",
        r"\bstomach pain\b",
        r"\babdominal pain\b",
        r"\bdiarrhea\b"
    ],

    "Neurological or General Symptoms": [
        r"\bheadache\b",
        r"\bdizziness\b",
        r"\bfatigue\b",
        r"\bweakness\b"
    ]

}

# Guidance associated with matched patterns

guidance = {

    "Emergency Warning Signs":
        "Seek emergency medical care immediately. Do not delay care "
        "while using this system.",

    "Respiratory Symptoms":
        "Consider professional medical advice if symptoms persist, "
        "worsen, or are accompanied by breathing difficulty.",

    "Digestive Symptoms":
        "Maintain hydration if appropriate and seek medical advice "
        "for persistent or worsening symptoms.",

    "Neurological or General Symptoms":
        "Consider medical assessment if symptoms persist or worsen. "
        "Sudden weakness or other sudden neurological changes require "
        "emergency assessment."

}


def match_symptoms(patient_input):

    text = patient_input.lower()

    matched_categories = []
    matched_symptoms = []

    for category, patterns in symptom_patterns.items():

        category_matches = []

        for pattern in patterns:

            if re.search(pattern, text):
                category_matches.append(
                    re.search(pattern, text).group()
                )

        if category_matches:
            matched_categories.append(category)
            matched_symptoms.extend(category_matches)

    return matched_categories, matched_symptoms


def healthcare_assistant():

    print("HEALTHCARE SYMPTOM ASSESSMENT SYSTEM")
    print("-" * 40)

    print("Please enter your symptoms:")
    print("Example: Fever, cough, headache and sore throat.")

    patient_input = input("\nEnter your symptoms: ").strip()

    if not patient_input:

        print("System: Please enter a symptom description.")
        return

    categories, symptoms = match_symptoms(patient_input)

    print("\nASSESSMENT REPORT")
    print("-" * 40)

    if not categories:

        print("Matched symptoms: None")
        print("System: No predefined pattern was matched.")

        print("Please describe the symptoms more clearly or "
              "consult a healthcare professional.")

        return

    print("Matched symptoms:", ", ".join(sorted(set(symptoms))))
    print("Matched categories:", ", ".join(categories))

    # Emergency patterns take priority over other matches

    if "Emergency Warning Signs" in categories:

        print("\nPriority: EMERGENCY WARNING SIGN")
        print("Guidance:", guidance["Emergency Warning Signs"])

    else:

        print("\nPriority: NON-EMERGENCY PATTERN MATCH")

        for category in categories:
            print(f"{category}: {guidance[category]}")

    print("\nNote: This system does not diagnose diseases "
          "or replace professional medical assessment.")


healthcare_assistant()
```
